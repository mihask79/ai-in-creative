## Независимая проверка файлов и метрик

```python
import json, hashlib
import lab05_lib as L
for name in ['run_A', 'run_B', 'error_A', 'error_B', 'fixed']:
    folder = ROOT / 'artifacts' / name
    m = json.loads((folder / 'manifest.json').read_text(encoding='utf-8'))
    assert hashlib.sha256((folder / 'result.png').read_bytes()).hexdigest() == m['sha256']
    with Image.open(folder / 'result.png') as im:
        assert im.size == (512, 512) and im.mode == 'RGB'
for key, names in {'repeat':('run_A','run_B'), 'shared_generator':('error_A','error_B'),
                   'repaired_vs_reference':('run_A','fixed')}.items():
    a, b = [Image.open(ROOT / 'artifacts' / n / 'result.png') for n in names]
    assert L.compare(a,b) == results[key]
    print(key, json.dumps(results[key], ensure_ascii=False))
print('PASS: SHA-256, размеры и метрики пересчитаны по сохранённым PNG')
```
